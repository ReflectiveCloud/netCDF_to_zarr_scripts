```python
import json
import gc
import os

import dask
import s3fs
import xarray as xr
import zarr

fs = s3fs.S3FileSystem(anon=False)

from dask.distributed import Client

import warnings
import logging

# Suppress zarr vlen-utf8 codec warning
warnings.filterwarnings("ignore", message=".*vlen-utf8.*")

# Suppress dask distributed shuffle warning
logging.getLogger("distributed.shuffle._scheduler_plugin").setLevel(logging.ERROR)

client = Client(n_workers=4, threads_per_worker=2, memory_limit="6GB")
print(client.dashboard_link)

# Helpers

def open_variable(table_dir, var, experiment, member):
    """Open a single NetCDF file, renaming the data variable to match var if needed."""
    path = f"{BASE_URL}/{table_dir}/{var}_{experiment}_{member}.nc"
    ds = xr.open_dataset(path, engine="h5netcdf", chunks={},
                         storage_options={"anon": False})
    data_vars = [v for v in ds.data_vars if not v.endswith("_bnds")]
    if len(data_vars) == 1 and data_vars[0] != var:
        ds = ds.rename({data_vars[0]: var})
    return ds


def _patch_zarr_json(fs, bucket_path, rel_path, dims):
    """
    Write dimension_names (zarr v3 native) and _ARRAY_DIMENSIONS (xarray legacy)
    directly into a zarr.json on S3, ensuring both fields are always in sync.
    """
    zarr_json_path = f"{bucket_path}/{rel_path}/zarr.json"
    with fs.open(zarr_json_path, "r") as f:
        meta = json.load(f)
    meta["dimension_names"] = dims
    meta.setdefault("attributes", {})["_ARRAY_DIMENSIONS"] = dims
    with fs.open(zarr_json_path, "w") as f:
        json.dump(meta, f)


def _write_var_to_zarr(combined, var, store_url, zarr_group_path, chunks, first_write):
    """
    Write a single variable to a zarr group.
    - First variable: use xarray to write coordinates, then patch their dimension
      metadata since xarray doesn't reliably set dimension_names in zarr v3.
    - All data variables: write via raw zarr so we fully control dimension metadata,
      bypassing xarray's coordinate validation which causes conflicts across groups.
    """
    group_url = f"{store_url}/{zarr_group_path}"
    bucket_path = store_url.replace("s3://", "")
    storage_options = {"anon": False}

    if first_write:
        # Write coordinates only via xarray
        coords_only = combined.drop_vars(var)
        coords_only.to_zarr(
            group_url,
            mode="w",
            consolidated=False,
            storage_options=storage_options,
        )
        # Patch dimension_names on all coordinates — xarray doesn't set this in zarr v3
        for coord_name in coords_only.coords:
            _patch_zarr_json(fs, bucket_path, f"{zarr_group_path}/{coord_name}", [coord_name])

    # Always write the data variable via raw zarr
    root = zarr.open_group(group_url, mode="a", storage_options=storage_options)
    da = combined[var]
    dims = list(da.dims)
    chunk_tuple = tuple(chunks.get(d, s) for d, s in zip(dims, da.shape))

    if var not in root:
        root.require_dataset(
            var,
            shape=da.shape,
            chunks=chunk_tuple,
            dtype=da.dtype,
            dimension_names=dims,
        )

    # Write data chunk-by-chunk via dask
    da.data.store(root[var], lock=False)

    # Patch zarr.json to ensure both dimension_names and _ARRAY_DIMENSIONS are set
    _patch_zarr_json(fs, bucket_path, f"{zarr_group_path}/{var}", dims)


def ensure_root_markers(store_path, table_names):
    """
    Write zarr.json group markers at the root and each table level.
    xarray only writes markers at leaf group level, so consolidate_metadata
    fails without these intermediate markers.
    """
    bucket_path = store_path.replace("s3://", "")
    marker = {"zarr_format": 3, "node_type": "group", "attributes": {}}
    for p in [""] + table_names:
        key = f"{bucket_path}/{p}/zarr.json" if p else f"{bucket_path}/zarr.json"
        with fs.open(key, "w") as f:
            json.dump(marker, f)
        print(f"  Wrote marker: {key}")


# Processing functions

def process_group(store_path, table_name, group_name, config, experiment, members):
    zarr_group_path = f"{table_name}/{group_name}"
    variables = config["variables"]
    chunks = config["chunks"]

    print(f"  Processing {zarr_group_path} [{experiment}]...")

    first_write = True
    for var in variables:
        print(f"    Variable: {var}")

        member_datasets = []
        for m in members:
            ds = open_variable(table_name, var, experiment, m)
            drop = [v for v in ds.data_vars if v.endswith("_bnds")]
            ds = ds.drop_vars(drop).expand_dims({"member": [m]})
            member_datasets.append(ds)

        combined = xr.concat(member_datasets, dim="member").chunk(chunks)
        _write_var_to_zarr(combined, var, store_path, zarr_group_path, chunks, first_write)
        first_write = False

        del combined, member_datasets
        gc.collect()

    print("    ✓ Written")


def process_day_group(store_path, group_name, config, experiment, members):
    zarr_group = f"day/{group_name}"
    variables = config["variables"]
    chunks = config["chunks"]

    print(f"  Processing {zarr_group} [{experiment}]...")

    first_write = True
    for var in variables:
        print(f"    Variable: {var}")

        member_datasets = []
        for m in members:
            ds = open_variable("day", var, experiment, m)
            drop = [v for v in ds.data_vars if v.endswith("_bnds")]
            ds = ds.drop_vars(drop).expand_dims({"member": [m]})
            member_datasets.append(ds)

        combined = xr.concat(member_datasets, dim="member").chunk(chunks)
        _write_var_to_zarr(combined, var, store_path, zarr_group, chunks, first_write)
        first_write = False

        del combined, member_datasets
        gc.collect()

    print("    ✓ Written")


# Main loop
print(tables)
for exp, store_path in STORES.items():
    print(f"\n{'='*50}")
    print(f"Writing store: {store_path}")
    print(f"{'='*50}")

    for table_name, table_config in tables.items():
        if exp not in table_config["experiments"]:
            continue

       
        for group_name, config in table_config["groups"].items():
            members = config["members"]
            if table_name == "day":
                process_day_group(store_path, group_name, config, exp, members)
            else:
                process_group(store_path, table_name, group_name, config, exp, members)

    ensure_root_markers(store_path, list(tables.keys()))
    zarr.consolidate_metadata(store_path)
    print(f"\n✓ Consolidated metadata for {store_path}")

print("\nAll done!")
```

```python
# Tool to consolidate metadata to allow the zarr store to function properly in Cloudflare R2.
# Use with the MIROC-ES2H stores and the CESM2-WACCM stores.
# Must be updated for each new store.

import s3fs, zarr
from zarr.storage import FsspecStore

fs = s3fs.S3FileSystem(
    key=os.environ.get("R2_ACCESS_KEY", None),
    secret=os.environ.get("R2_SECRET_KEY", None),
    endpoint_url=f"https://{os.environ.get("R2_ACCOUNT_ID", None)}.r2.cloudflarestorage.com",
    client_kwargs={'region_name': 'auto'},
)
store = FsspecStore(fs, path='reflective-data-store/CESM2-WACCM/CESM2-WACCM_SSP2-4.5.zarr')

subgroups = [
    None,                      # root (already been done)
    'Amon/atmos_2d',
    'Amon/atmos_3d',
    'day/atmos_2d',
    'Lday',
    'Lmon/land_2d',
    'Omon/ocean_2d',
    'Omon/ocean_3d',
    'Oyr',
    'SIday/seaice_2d',
    'SImon/seaice_2d'
]

for path in subgroups:
    print(f'Consolidating {path or "<root>"}...')
    zarr.consolidate_metadata(store, path=path)

print('Done.')
```
